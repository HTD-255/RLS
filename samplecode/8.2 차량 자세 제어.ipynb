```python
from pop import Pilot
import numpy as np
import time

Car = Pilot.AutoCar()
Car.setSensorStatus(euler=1)
```

```python
dataset={ 'gyro' : [], 'euler' : [], 'steer' : [] }

for n in np.arange(-0.9, 1.1, 0.3):
    n = round(n,1)

    Car.steering = n
    Car.forward()

    time.sleep(1)

    gy = Car.getGyro('z')
    eu = Car.getEuler('yaw')

    time.sleep(1)

    Car.backward()

    time.sleep(2)

    Car.stop()

    n = -n    # 조향 제어값을 반전하여 저장

    dataset['gyro'].append( gy )
    dataset['euler'].append( eu )
    dataset['steer'].append( n ) 

    print({ 'gyro' : gy, 'euler' : eu , 'steer' : n }) 
```

```python
from pop import AI

LR=AI.Linear_Regression()
LR.X_data=dataset['gyro']
LR.Y_data=dataset['steer']

LR.train(times=5000, print_every=100)
```

```python
LR.run()
```

```python
Car.forward()

while True:
    try:
        err = Car.getGyro('z')
        value = LR.run([err])[0][0]

        Car.steering = value * 1.5
        time.sleep(0.1)
    except KeyboardInterrupt:
        Car.stop()
        break
```
