```python
import time
from pop import Pilot

Car = Pilot.AutoCar()
```

```python
Car.steering = -1

time.sleep(1)

Car.steering = 1
```

```python
Car.forward()

time.sleep(1)

Car.backward()

time.sleep(1)

Car.stop()
```

```python
Car.setSpeed(50)
Car.forward()

time.sleep(1)

Car.backward(99)

time.sleep(1)

Car.stop()
```

```python
Car.joystick()
```

```python
Car.setLamp(1, 0)
time.sleep(1)

Car.setLamp(1, 1)
time.sleep(1)

Car.setLamp(0, 0)
```

```python
Car.setCalibration()
```

```python
value = Car.getAccel()
print(value)

time.sleep(1)

value = Car.getAccel('z')
print(value) 
```

```python
value = Car.getMagnetic()
print(value)

time.sleep(1)

value = Car.getMagnetic('z')
print(value) 
```

```python
value = Car.getGyro()
print(value)

time.sleep(1)

value = Car.getGyro('z')
print(value) 
```

```python
Car.setSensorStatus(euler=1)

value = Car.getEuler()
print(value)

time.sleep(1)

value = Car.getEuler('yaw')
print(value) 
```

```python
Car.setSensorStatus(gravity=1)
time.sleep(0.5)

value = Car.getGravity()
print(value)

time.sleep(1)

value = Car.getGravity('z')
print(value) 
```
