```python
import csv, time
from pop import Pilot
Car = Pilot.AutoCar()
```

```python
ultrasonic_front = []
for _ in range(50):
    val = Car.getUltrasonic()
    print(val)
    ultrasonic_front.append(val[0])
    
    time.sleep(0.5)
```

```python
with open("ultra.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(map(lambda x: [x], ultrasonic_front))
```

```python
with open('ultra.csv','r') as csv_file:
    lines = csv_file.readlines()

print(lines)
```

```python
ultra = []
score = []
for line in lines:
    tmp = line.split(",")
    ultra.append([int(tmp[0])])
    score.append([int(tmp[1].split("\n")[0])])

print(ultra[0], score[0])
```

```python
from pop import AI

Perc = AI.Perceptron(input_size=1)
Perc.X_data = ultra
Perc.Y_data = score
```

```python
Perc.train()
```

```python
val = Car.getUltrasonic()[0]
result = Perc.run([val])
val, result[0][0]
```

```python
for _ in range(100):
    val = Car.getUltrasonic()[0]
    result = Perc.run([val])[0][0]
    print({"Ultrasonic" : val, "Result" : result})
    if result > 0.3: 
        Car.alarm(8,8,0.5)
```
