```python
from pop import Dataset

Dataset.Collector("Audio")
```

```python
import os
import numpy as np
from pop import AI
from pop import Util

X_data=[]
Y_data=[]

datalist = os.listdir("audio_dataset")

for data in datalist:
    feat = Util.toMFCC("audio_dataset/" + data, duration=1)
    label = int(data.split("_")[0])
    label = Util.one_hot(label,10)
    
    X_data.append(feat)
    Y_data.append(label)

X_data=np.array(X_data)
Y_data=np.array(Y_data)
```

```python
dataset_size = X_data.shape[1:3]
CNN=AI.CNN(input_size=dataset_size, output_size=10)

CNN.X_data=X_data
CNN.Y_data=Y_data

CNN.train(times=100)
```

```python
Y=CNN.Y_data[20]
R=CNN.run([CNN.X_data[20]])

print(Y)
print(R)
```
