```python
from pop import LiDAR

lidar = LiDAR.Rplidar()
lidar.connect() 
lidar.startMotor()
```

```python
vectors = lidar.getVectors()

for v in vectors:
    print(v)
```

```python
coords = lidar.getXY()

for c in coords:
    print(c)
```
