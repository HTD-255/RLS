```python
import cv2
import subprocess
from popAssist import *
from pop import Camera

cam = Camera(width=300, height=300)

def userAction(text): 
    action = False

    r = text.find("사진")
    if r != -1 and text.find("찍어", r) != -1:
        print("Take picture...")
        cv2.imwrite("picture.png", cam.value)
        action = True

    return action

stream = create_conversation_stream()
ga = GAssistant(stream, local_device_handler=userAction) 

try:
    def onStart():
        print(">>> Start recording....")

    while True: 
        ga.assist(onStart)
except KeyboardInterrupt:
    stream.close()
```

```python
from pop import Pilot
from popAssist import *

Car = Pilot.AutoCar()
```

```python
def userAction(text): 
    action = False
    
    print(text)

    if text.find("전진") != -1:
        Car.forward(30)
        action = True
    elif text.find("후진") != -1:
        Car.backward(30)
        action = True
    elif text.find("정지") != -1:
        Car.stop()
        action = True

    return action
```

```python
stream = create_conversation_stream()
ga = GAssistant(stream, local_device_handler=userAction) 

try:    
    def onStart():
        print(">>> Start recording....")

    while True: 
        ga.assist(onStart)
except KeyboardInterrupt:
    stream.close()
```
