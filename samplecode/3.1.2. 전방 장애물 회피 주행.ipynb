```python
from pop import LiDAR, Pilot

lidar = LiDAR.Rplidar()
car = Pilot.AutoCar()
```

```python
lidar.connect()
lidar.startMotor()

car.setSpeed(50)
car.steering = 0
```

```python
while True:
    no_collision = True
    
    vectors = lidar.getVectors()
    for v in vectors:
        degree = v[0]
        distance = v[1]
        
        if degree <= 60 or degree >= 300:
            if distance <= 700: 
                no_collision = False
                
    if no_collision:
        car.setSpeed(50)
        car.forward()
    else:
        if car.steering == 0:
            car.stop()
        car.steering = -1
        car.backward(30)
```
