```python
from tensorflow import keras

model = keras.models.Sequential()

model.add(keras.layers.InputLayer(input_shape=(1,)))
model.add(keras.layers.Dense(50))
model.add(keras.layers.Dense(1))

value = model.predict([1])
print(value)
```

```python
model.summary()
```

```python
model.compile(loss='MSE', optimizer='SGD')
```

```python
from tensorflow import keras

model = keras.models.Sequential()

model.add(keras.layers.InputLayer(input_shape=(1,)))
model.add(keras.layers.Dense(1))

model.compile(loss='MSE', optimizer='SGD')

X=[[0],[1],[2],[3],[4],[5]]
Y=[[0],[2],[4],[6],[8],[10]]

model.fit(X, Y, epochs=100)
```

```python
value = model.predict([[7], [-10], [358]])
print(value)
```

```python
from tensorflow import keras

model = keras.models.Sequential()

model.add(keras.layers.InputLayer(input_shape=(1,)))
model.add(keras.layers.Dense(1, activation='sigmoid'))

model.compile(loss='binary_crossentropy', optimizer='Adam')

X=[[-3],[-2],[-1],[1],[2],[3]]
Y=[[0],[0],[0],[1],[1],[1]]

model.fit(X, Y, epochs=100)
```

```python
value = model.predict([[7], [-10], [358]])
print(value)
```

```python
from tensorflow import keras

model = keras.models.Sequential()

model.add(keras.layers.InputLayer(input_shape=(3,)))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dense(2, activation='sigmoid'))

model.compile(loss='binary_crossentropy', optimizer='Adam')

X=[[1,-2,3],[2,-3,4],[3,5,4],[4,-5,6],[7,-5,3], [1,6,8],[3,8,1]]
Y=[[0, 1],[0, 1],[1, 0],[0, 1],[1, 1],[0, 1],[1, 1]]

model.fit(X, Y, epochs=20)
```

```python
value = model.predict([[10,-5,3], [-1,6,-8], [3,5,8]])
print(value)
```
