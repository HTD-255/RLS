```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()

W = tf.Variable(1.)
B = tf.Variable(0.)
X = tf.placeholder(tf.float32)
Y = tf.placeholder(tf.float32)
```

```python
H = W * X + B
Loss = tf.reduce_mean(tf.reduce_mean(tf.abs(Y - H)))
```

```python
optimizer = tf.train.GradientDescentOptimizer(learning_rate=0.001)
train_op = optimizer.minimize(Loss)
```

```python
sess = tf.Session()
sess.run(tf.global_variables_initializer())
```

```python
X_data = [[0],[1],[2],[3],[4]]
Y_data = [[0],[2],[4],[6],[8]]

for i in range(100):
    sess.run(train_op, feed_dict={X : X_data, Y : Y_data})
    loss = sess.run(Loss, feed_dict={X : X_data, Y : Y_data})
    print(loss)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
 
X = tf.placeholder(tf.float32)
Y = tf.placeholder(tf.float32)
W1 = tf.Variable(tf.random_uniform([1, 10], -1., 1.))
W2 = tf.Variable(tf.random_uniform([10, 1], -1., 1.))
```

```python
L = tf.matmul(X, W1)     # input – hidden
L = tf.nn.relu(L)
model = tf.matmul(L, W2) # hidden - output
Loss = tf.reduce_mean(tf.reduce_mean(tf.square(Y - model)))
```

```python
optimizer = tf.train.GradientDescentOptimizer(learning_rate=0.001)
train_op = optimizer.minimize(Loss)
```

```python
sess = tf.Session()
sess.run(tf.global_variables_initializer())
```

```python
X_data = [[-4],[-3],[-2],[-1],[0],[1],[2],[3],[4]]
Y_data = [[-8],[-6],[-4],[-2],[0],[2],[4],[6],[8]] 

for i in range(500):
    sess.run(train_op, feed_dict={X : X_data, Y : Y_data})
    loss = sess.run(Loss, feed_dict={X : X_data, Y : Y_data})
    print(loss)
```

```python
n = [[7], [-10], [358]]
result = sess.run(model, feed_dict={X : n})
print(result)
```

```python
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()

X = tf.placeholder(tf.float32)
Y = tf.placeholder(tf.float32)
W1 = tf.Variable(tf.random_uniform([1, 10], -1., 1.))
W2 = tf.Variable(tf.random_uniform([10, 10], -1., 1.))
W3 = tf.Variable(tf.random_uniform([10, 1], -1., 1.))
```

```python
L1 = tf.matmul(X, W1)     # input – hidden
L1 = tf.nn.relu(L1)
L2 = tf.matmul(L1, W2)    # hidden – hidden
L2 = tf.nn.relu(L2)
model = tf.matmul(L2, W3) # hidden - output
Loss = tf.reduce_mean(tf.reduce_mean(tf.square(Y - model)))
```

```python
optimizer = tf.train.GradientDescentOptimizer(learning_rate=0.001)
train_op = optimizer.minimize(Loss)
```

```python
sess = tf.Session()
sess.run(tf.global_variables_initializer())
```

```python
X_data = [[0],[1],[2],[3],[4]]
Y_data = [[0],[2],[4],[6],[8]]

for i in range(500):
    sess.run(train_op, feed_dict={X : X_data, Y : Y_data})
    loss = sess.run(Loss, feed_dict={X : X_data, Y : Y_data})
    print(loss)
```

```python
n = [[7], [-10], [358]]
result = sess.run(model, feed_dict={X : n})
print(result)
```
