```python
import pandas as pd

value = [[7, 5],
         [5, 1],
         [3, -1]]
df = pd.DataFrame(value)

value2 = [[1, -3]]
df2 = pd.DataFrame(value2) 

print(df.add(df2))
print(df.sub(df2))
print(df.mul(df2))
print(df.div(df2))
```

```python
print(df.add(df2, fill_value=0))
print(df.sub(df2, fill_value=0))
print(df.mul(df2, fill_value=1))
print(df.div(df2, fill_value=1))
```

```python
import pandas as pd

value = [[7, 5],
         [5, 1],
         [3, -1]]
df = pd.DataFrame(value)

print(df.sum(0))
print(df.mean(0))
print(df.std(0))
print(df.var(0))
```
