```python
from pop import AI, Pilot
import time
       
LR = AI.Linear_Regression()
car = Pilot.AutoCar()
car.setSensorStatus(cds=1)
```

```python
X_data_1 = []
X_data_2 = []
Y_data_1 = []
Y_data_2 = []
```

```python
for _ in range(20):
    val = car.getLight()
    X_data_1.append(val)
    Y_data_1.append(1)
    print(val, end = " ")
    time.sleep(1)
```

```python
for _ in range(20):
    val = car.getLight()
    X_data_2.append(val)
    Y_data_2.append(0)
    print(val, end = " ")
    time.sleep(1)
```

```python
LR.X_data = X_data_1 + X_data_2
LR.Y_data = Y_data_1 + Y_data_2
```

```python
LR.train(times=500, print_every=50)
```

```python
LR.run()
```

```python
log = time.time()
while time.time()-log <= 20:
    predict = LR.run(car.getLight())
    if predict >= 0.5:
        car.setLamp(0,0)
    else:
        car.setLamp(1,1)
```
