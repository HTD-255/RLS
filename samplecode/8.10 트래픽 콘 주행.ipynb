```python
Car=Pilot.AutoCar()
Car.setSpeed(50)

def drive(value):
    Car.forward()
    steer=value['x']

    if steer > 1:
        steer=1
    elif steer < -1:
        steer=-1

    Car.steering=steer*1.5

while True:
    CF.run(callback=drive)
```
