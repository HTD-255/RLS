```python
from pop import AI, Camera, Pilot

cam=Camera(width=300, height=300)

OF=AI.Object_Follow(cam)
OF.load_model()
```

```python
v=OF.detect(index='person')
print(v)
```

```python
v=OF.detect()
print(v)
```

```python
OF.show()
```

```python
Car = Pilot.AutoCar()

while True:
    v=OF.detect(index='person')
 
    if v is not None:
        steer=v['x']*4

        if steer > 1:
            steer=1
        elif steer < -1:
            steer=-1

        Car.steering=steer

        if v['size_rate']<0.20:
            Car.forward(50)
        else:
            Car.stop()
    else:
        Car.stop()
```
