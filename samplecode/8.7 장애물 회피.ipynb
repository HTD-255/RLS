```python
def is_blocked(value):
    print(value>0.5)

value=CA.run(callback=is_blocked)
print(value)
```

```python
Car = Pilot.AutoCar()
Car.setSpeed(50)

def drive(value):
    if value<=0.5:
        Car.steering=0
        Car.forward()
    else:
        Car.steering=1
        Car.backward()

while True:
    CA.run(callback=drive)
```
